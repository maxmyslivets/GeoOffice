```python
from pathlib import Path
from datetime import datetime

def scan_dir(path: str | Path) -> tuple[datetime, datetime]:
    """
    Сканирует указанную папку и возвращает кортеж с минимальным и максимальным
    временем фотографий (по времени последней модификации файла).

    :param path: путь к папке для сканирования
    :return: (время первой фотографии, время последней фотографии)
    :raises FileNotFoundError: если папка не существует
    :raises ValueError: если не найдено ни одной фотографии
    """
    path = Path(path)
    if not path.exists():
        raise FileNotFoundError(f"Папка {path} не найдена")

    patterns = ["*.jpg", "*.jpeg", "*.JPG", "*.JPEG"]

    min_time, max_time = None, None

    for pattern in patterns:
        for item in path.rglob(pattern):
            if item.is_file():
                dt = datetime.fromtimestamp(item.stat().st_mtime)

                if min_time is None or dt < min_time:
                    min_time = dt
                if max_time is None or dt > max_time:
                    max_time = dt

    if min_time is None:
        raise ValueError(f"В папке {path} нет фотографий")

    return min_time, max_time

def format_datetime(dt: datetime) -> dict[str, str]:
    """
    Форматирует объект datetime в словарь с отдельными полями даты и времени.

    :param dt: объект datetime
    :return: словарь вида {"date": "дд.мм.гггг", "time": "чч:мм:сс"}
    """
    return {
        "date": dt.strftime("%d.%m.%Y"),
        "time": dt.strftime("%H:%M:%S"),
    }

paths = [
    r"\\server-r\IGI\Work\Объекты\2025\Полоцк\NN.GG ПДП Аэропорт\Полёты\Исходные\DJI_202509261537_002_cherneschino-аэропорт-1",
    r"\\server-r\IGI\Work\Объекты\2025\Полоцк\NN.GG ПДП Аэропорт\Полёты\Исходные\DJI_202509261603_003_cherneschino-аэропорт-8",
    r"\\server-r\IGI\Work\Объекты\2025\Полоцк\NN.GG ПДП Аэропорт\Полёты\Исходные\DJI_202509261818_003_cherneschino-аэропорт-8-copy",
    r"\\server-r\IGI\Work\Объекты\2025\Полоцк\NN.GG ПДП Аэропорт\Полёты\Исходные\DJI_202509261845_005_cherneschino-аэропорт-8-copy-copy",
]
for dir in paths:
    print(dir)
    times = scan_dir(dir)
    start, end = format_datetime(times[0]), format_datetime(times[1])
    print(f"{start['date']}\t{start['time']}\t{end['time']}")
```
